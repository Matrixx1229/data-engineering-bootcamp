# Basics

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Spark DataFrames").getOrCreate()
print("spark session created :", spark.version)
```

## Understanding what dataframe is

```python
#Create spark data frames

data= [('Anurag',50000),('Priya',67000),('Ravi',45000),('Stuti',90000),('Ankur',40000)]
coloumn= ['Name','Salary']

spark_df = spark.createDataFrame(data, coloumn)
display(spark_df)
```

```python
# pandas df

import pandas as pd
data= [('Anurag',50000),('Priya',67000),('Ravi',45000),('Stuti',90000),('Ankur',40000)]
coloumn= ['Name','Salary']
pandas_df = pd.DataFrame(data, columns=coloumn)
display(pandas_df)
```

```python
spark_df.printSchema()
```

```python
print("Counts in df", spark_df.count())
print("Counts in pandas df", len(pandas_df))
```

### Real-Life Use Case
These Dataframes are used for :
- Ingesting data from files
- Transforming at scale
- writing to delta tables or parquet files

```python
# add data types

from pyspark.sql.types import StructType,StructField, StringType, IntegerType
schema = StructType([ \
    StructField("Name", StringType(), True), \
    StructField("Salary", IntegerType(), True) \
  ])
spark_df = spark.createDataFrame(data, schema=schema)
display(spark_df)
spark_df.printSchema()
```

```python
json_data = [
    {"Name": "Anurag", "Salary": 50000},
    {"Name": "Priya", "Salary": 67000}
]

df_json = spark.createDataFrame(json_data)
display(df_json)
```
