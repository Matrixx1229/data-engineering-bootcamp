## Layers in Architecture (Medallion):
Bronze ➝ Raw data
Silver ➝ Cleaned + enriched data
Gold ➝ Aggregated insights

### Build Medallion Architecture (Practical)
🔶 Bronze – Ingest Raw CSV

```python
df_raw = spark.read.option("header", True).csv("/databricks-datasets/retail-org/customers/")
df_raw.write.format("delta").mode("overwrite").save("/Volumes/workspace/default/datax/lakehouse/bronze/customers")
display(df_raw)
```

## 🔷 Silver - Clean and Standard

```python
# Silver - Clean and Standard 
df_bronze = spark.read.format("delta").load("/Volumes/workspace/default/datax/lakehouse/bronze/customers")

# Display the schema to check the column names
df_bronze.printSchema()

```

```python
from pyspark.sql.functions import col, lower, trim

# Assuming the correct column name is 'customer_email' instead of 'email'
df_silver = df_bronze.withColumn("customer_name", lower(trim(col("customer_name")))) \
                     .dropna(subset=["customer_id", "customer_name"])

df_silver.write.format("delta").mode("overwrite").save("/Volumes/workspace/default/datax/lakehouse/silver/customers")

display(df_silver)
```

## 🟡 Gold – Aggregated Insights

```python
df_silver = spark.read.format("delta").load("/Volumes/workspace/default/datax/lakehouse/silver/customers")

df_gold = df_silver.groupBy("state").count()
df_gold.write.format("delta").mode("overwrite").saveAsTable("customer_summary_gold")
```
