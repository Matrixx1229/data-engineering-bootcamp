DDL Schema

```python
my_ddl_schema = """
                    Item_Identifier STRING,
                    Item_Weight STRING,
                    Item_Fat_Content STRING, 
                    Item_Visibility DOUBLE,
                    Item_Type STRING,
                    Item_MRP DOUBLE,
                    Outlet_Identifier STRING,
                    Outlet_Establishment_Year INT,
                    Outlet_Size STRING,
                    Outlet_Location_Type STRING, 
                    Outlet_Type STRING,
                    Item_Outlet_Sales DOUBLE 

            """
```

```python
df = (
    spark.read.format("csv")
    .schema(my_ddl_schema)
    .load("/Volumes/dataxbootcamp/default/datax/BigMart Sales.csv")
)
```

```python
df = (
    spark.read.format("csv")
    .option("inferSchema", True)
    .option("header", True)
    .load("/Volumes/dataxbootcamp/default/datax/BigMart Sales.csv")
)

df.display()
```

## FILTER

SCENE - 1

```python
df.filter(col('Item_Fat_Content') == 'Regular').display()
```

SCENE - 2

```python
ds = df.filter((col('Item_Weight')<10) & (col('Item_Type') == "Soft Drinks"))
display(ds)
```

SCENE - 3

```python
df.filter((col('Outlet_Size').isNull()) & (col('Outlet_Location_Type').isin('Tier 1','Tier 2'))).display()
```

### withColumnRenamed

```python
df.withColumnRenamed('Item_Weight', 'Item_WT').display()
```

### withColumn

```python
# SCENE - 1

df = df.withColumn('Flag', lit('New'))
df.display()
```

```python
df.withColumn('Multiply', col('Item_Weight')*col('Item_MRP')).display()
```

```python
# SCENE 2

df.withColumn('Item_Fat_Content', regexp_replace(col('Item_Fat_Content'), 'Low Fat', 'LF'))\
    .withColumn('Item_Fat_Content', regexp_replace(col('Item_Fat_Content'), 'Regular', 'Reg')).display()
```

### Type Casting

```python
df.withColumn('Item_Weight', col('Item_Weight').cast(DoubleType())).printSchema()
```
