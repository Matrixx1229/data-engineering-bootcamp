## Inheritence (Single, Multilevel, Multiple)

```python
class company():
    def company_info(self):
        print("Compnay name is XYZ")

class employee(company):
    def emp_info(self):
        print("EMployee name onkar")

emp1 = employee()
emp1.emp_info()
emp1.company_info()
```

```python
class company():
    def __init__(self, com_name):
        self.com_name = com_name

    def company_info(self):
        print("Compnay name is XYZ")

class employee(company):
    def __init__(self,emp_name, com_name):
        self.emp_name = emp_name
        self.com_name = com_name
        
    def emp_info(self):
        print("EMployee name onkar")

emp1 = employee("Onkar", "XYS")

emp1.emp_info()
emp1.company_info()
```

```python
class company():
    def __init__(self, com_name, country):
        self.com_name = com_name
        self.counrty = country

    def company_info(self):
        print(f"Compnay name is {self.com_name} and country is {self.counrty}")

class employee(company):
    def __init__(self,emp_name, com_name, country):
        self.emp_name = emp_name
        
        company.__init__(self, com_name, country)
        # self.com_name = com_name
        
    def emp_info(self):
        print("EMployee name onkar")
    
    def company_info_child(self):
        print("this is running from employee")
        company.company_info(self)
        # super().company_info()

emp1 = employee("Onkar", "XYS", "India")
emp1.company_info_child()
```

```python
class company():
    def __init__(self, com_name):
        self.com_name = com_name

    def company_info(self):
        print(f"Compnay name is {self.com_name}.")

class country():
    def __init__(self, country_name):
        self.country_name = country_name

    def country_info(self):
        print(f"Country name is {self.country_name}.")

class employee(company, country):
    def __init__(self,emp_name, com_name,country_name):
        self.emp_name = emp_name
        
        company.__init__(self, com_name)

        country.__init__(self, country_name)
        # self.com_name = com_name
        
    def full_info(self):
        print(f"The employee {self.emp_name} lives in {self.country_name} and works for {self.com_name}")
    
    def all_info_child(self):
        print("this is running from employee")
        company.company_info(self)
        country.country_info(self)
        # super().company_info()

emp1 = employee("Onkar", "XYS", "India")
emp1.full_info()

emp1.all_info_child()
```
