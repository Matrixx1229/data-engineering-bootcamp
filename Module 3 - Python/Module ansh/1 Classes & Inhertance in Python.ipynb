```python
class llms:

    token_size = 100

    def __init__(self,query):
        self.query = query

    def openai(self):
        print(f"i am openai. You asked {self.query}")

    def claude(self):
        print("i am claude")

    def llama(self):
        print("i am llama")

obj = llms("Hey, who is ansh lamba. I know he is amazing but give me an answer.")
obj.openai()
```

```python
# from Module_3_-_Python/Module_ansh/advanced_python_1 import llms
class llms:

    token_size = 100

    def __init__(self,query):
        self.query = query

    def openai(self):
        print(f"i am openai. You asked {self.query}")

    def claude(self):
        print("i am claude")

    def llama(self):
        print("i am llama")
  
if __name__ == "__main__":

  obj = llms("ansh is ansh")
  print(obj.openai())

# from firstClass import llms
class chatbot(llms):
    def __init__(self, model, query):
        self.model = model
        llms.__init__(self, query)

    def showme(self):
        print(f"i am calling{self.model}")
        llms.claude()

obj_inhert = chatbot("openai", "Hey, I am Onkar Bele")
print(obj_inhert.openai())
```
