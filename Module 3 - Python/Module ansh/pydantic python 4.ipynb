```python
from pydantic import BaseModel, Field, StrictInt
```

```python
class input(BaseModel):
    x : StrictInt = Field(..., description = "this is x")
    y : str = Field(..., description = "this is y")

# pyd_input = input(x=10, y = "Onkar Bele")
pyd_input = input(**{"x":10, "y":"Onkar Bele"})

https://dbc-d6aa71de-d3fe.cloud.databricks.com/editor/notebooks/348310731204826?o=3253117912920111$0print(pyd_input)
```

```python
class onkar(BaseModel):
    x : StrictInt = Field(..., description = "this is x")
    y : str = Field(..., description = "this is y")

pyd_input = onkar(**{"x":10, "y":"Onkar Bele"})

def main(para1:onkar):
    print("hello onkar bele")
main(pyd_input)
```

```python
class onkar(BaseModel):
    x : StrictInt = Field(..., description = "this is x")
    y : str = Field(default="Onkar bele")

pyd_input = onkar(**{"x":10})

def main(para1:onkar):
    return para1.y
main(pyd_input)
```
