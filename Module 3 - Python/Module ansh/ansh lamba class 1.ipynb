```python
my_list=[1,2,3,4,5,6]

# for i in reversed (my_list):
#   print(i)

my_list.reverse()
print(my_list)
```

```python
my_list=[1,2,3,4,5,6]

print(my_list[::-1])
```

```python
new_list = []
for i in my_list:
    new_list.append(i*i)
print(new_list)
```

```python
new_list = [i**2 for i in my_list]
print(new_list)

new_list = [i**2 for i in my_list if (i%2)== 0 ]
print(new_list)

new_list = [i**2 for i in my_list if (i%2)== 0 if (i !=6)]
print(new_list)
```

### Dictionary 

```python
my_dict = {"x": 1, "y": 2, "z": 3}
print(my_dict)
```

```python
my_dict["x"] = 4
my_dict.pop("z")
print(my_dict)
```

```python
print(my_dict.keys())
print(my_dict.values())
print(my_dict.items())
```

```python
my_dict = {"x": 1, "y": 2, "z": 3, "demo": {"a": 1, "b": 2, "c": 3}}
print(my_dict["demo"]["b"])
```

### Sets

```python
my_set_a = {1,2,3,4,5,5,5,5,5}
my_set_b = {10, 11, 3, 2, 5}
print(my_set_a.union(my_set_b))
```

```python
print(my_set_a.intersection(my_set_b))
```

```python
my_set_a.remove(4)
my_set_a.add(20)
print(my_set_a)
```

```python
a= {}
print(type(a))

b = set()
print(type(b))
```

### Tuple

```python
my_tup = (1,2,3,4,5)
print(my_tup)

my_tuplist = list(my_tup)
my_tuplist.append(6)
print(my_tuplist)

my_tup = tuple(my_tuplist)
print(my_tup)
```
