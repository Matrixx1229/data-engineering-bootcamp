## Map Filter Reduce

```python
my_list = [1,2,3,4,5]

def square(p_x):
  ret_list= [i*i for i in p_x]
  print(ret_list)
square(my_list)
```

```python
def square(p_x):

    return p_x*p_x

# result = list(map(square,my_list))
# print(result)
result = tuple(map(square,my_list))
print(result)
```

```python
from functools import reduce

my_list = [1,2,3,4,5]

def square(p_x, p_y):
  return p_x+p_y
  
result = reduce(square,my_list)
print(result)
```

## Execption Handling


```python
x = "10" 
try:
  if (x>10):
    print("x is greater than 10")
  else:
    print("x is less than 10")

except:
  print("hey you have an error")
print('hello world')
```

```python
x = 10 
try:
  if (x>10):
    print("x is greater than 10")
  else:
    print("x is less than 10")

except Exception as e:
    print(f"hey you have an error - {e}")
finally:
    print('I will always run')
```

```python
def my_func(p_x):
    try:
        if(p_x % 2 == 0):
            return 1
        else:
            return 0
    except Exception as e:
        print(f"hey you have an error - {e}")
my_func(20)
```
