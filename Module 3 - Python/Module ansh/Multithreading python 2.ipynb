```python
import threading 
from concurrent.futures import ThreadPoolExecutor
```

```python
def trasnformation(input):
    src = input['src']
    dst = input['dst']
    
    print(f"reading data from {src}")
    print(f"Transforming.....")
    print(f"writing data to {dst}")

array = [
    {"src":"table1","dst":"table1"},
    {"src":"table2","dst":"table2"},
    {"src":"table3","dst":"table3"}
]

with ThreadPoolExecutor(max_workers=3) as executor:
    my_futures = executor.map(trasnformation,array)
    
```

```python
def trasnformation(input):
    src = input['src']
    dst = input['dst']
    
    print(f"reading data from {src}")
    print(f"Transforming.....")
    print(f"writing data to {dst}")
    return f"Done {src} to {dst}"

array = [
    {"src":"table1","dst":"table1"},
    {"src":"table2","dst":"table2"},
    {"src":"table3","dst":"table3"}
]

with ThreadPoolExecutor(max_workers=3) as executor:
    my_futures = executor.map(trasnformation,array)
print(f"The return value is {list(my_futures)}")
```
