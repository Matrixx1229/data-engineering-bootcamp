```python
from pydantic import BaseModel, Field, EmailStr
from typing import Optional, List, Dict, Any, Literal

class personel_info(BaseModel):
    name : str = Field(..., min_length=3, max_length=20, description="Name of the person")
    age : int = Field(..., ge=0, description="Age of the person")
    email : EmailStr = Field(..., description="Email of the person")
    gender : Literal["male", "Female", "Other"] = Field(..., description="Gender of the person")
    salaries : List[int] = Field(..., description="Salaries of the person")
    
def main(para1: personel_info):
    print("name:", para1.name)
    print("age:", para1.age)
    print("email:", para1.email)
    print("gender:", para1.gender)
    print("salaries:", para1.salaries)

pyd_ins = personel_info(**{"name":"onkar bele", "age":21, "email":"onkarbele@gmail.com", "gender":"male", "salaries":[0,0]})
main(pyd_ins)
```

field validation 

```python
from pydantic import BaseModel, Field, field_validator
from typing import Optional, List, Dict, Any, Literal

class personal_info(BaseModel):
    name : str = Field(..., min_length=3, max_length=20, description="Name of the person")
    age : int = Field(..., ge=0, description="Age of the person")
    email : str = Field(..., description="Email of the person")

    @field_validator("email")
    def validate_email(cls, value):
      if "@" not in value or ".com" not in value:
        raise ValueError("Invalid  address email")
      return value

pyd_ins = personal_info(**{"name":"onkar bele", "age":21, "email":"onkarbele@gmail.com"})  
print(pyd_ins)  
```

```python
class personal_info(BaseModel):
    name : str = Field(..., min_length=3, max_length=20, description="Name of the person")
    age : int = Field(..., ge=0, description="Age of the person")
    email : str = Field(..., description="Email of the person")

    @field_validator("email")
    def validate_email(cls, value):
      if "@" not in value:
        raise ValueError("Invalid  address email")
      elif ".com" not in value:
        return value + ".com"
      else:
        return value

pyd_ins = personal_info(**{"name":"onkar bele", "age":21, "email":"onkarbele@gmail"})  
print(pyd_ins)  
```

MODEL VALIDATION


```python
from pydantic import BaseModel, Field, field_validator, model_validator
from typing import Optional, List, Dict, Any, Literal

class API_auth(BaseModel):
    api_key: str = Field(..., min_length=10, description="API key for authentication")
    api_secret: str = Field(..., min_length=10, description="API secret for authentication")
    environment: Literal["dev", "prod", "test"] = Field(..., description="Environment name")

    @model_validator(mode="after")
    def check_api_fields(cls, values):
        if values.api_key == values.api_secret:
            raise ValueError("api_key and api_secret must not be the same")
        return values
pyd_ins = API_auth(**{"api_key":"1234567890", "api_secret":"1234567898", "environment":"dev"})
print(pyd_ins)
```

Computed fields

```python
from pydantic import BaseModel, Field, computed_field
from typing import Optional, Dict, Any, List, Literal 


class orders(BaseModel):

    id : int = Field(...,description="Order Id")
    units: int = Field(..., description="Units")
    amount : int = Field(...,description="Amount Per Unit")

    @computed_field
    @property
    def total(self)->int:
        return self.units * self.amount

pyd_ins = orders(**{"id":1, "units":10, "amount":100})
print(pyd_ins)
```

Serialization

```python
from pydantic import BaseModel, Field, computed_field
from typing import Optional, Dict, Any, List, Literal 

class input(BaseModel):
    
    query : str = Field(..., description="This is query")

class output(BaseModel):
    
    query : str = Field(..., description="query")
    result : str = Field(..., description="Result")

def process_data(p_input:input) -> output:

    input_query = p_input.query
    result = "This is result"
    pyd_output = output(**{"query":input_query, "result":result})

    return pyd_output

input_query = input(**{"query":"This is query"})
response = process_data(input_query)
print(response)

# Serialization to Dictionary
print("-----------------------------------")
print(response.model_dump())

print("-----------------------------------")
print(response.model_dump_json())
```

Inheritance

```python
from pydantic import BaseModel, Field
from typing import Optional, Dict, Any, List, Literal

class Address(BaseModel):
    street: str = Field(..., description="Street")
    city: str = Field(..., description="City")
    state: str = Field(..., description="State")
    country: str = Field(..., description="Country")
    zip: str = Field(..., description="Zip")

class PersonalInfo(BaseModel):
    name: str = Field(..., min_length=3, max_length=20, description="Name of the person")
    age: int = Field(..., ge=0, description="Age of the person")
    email: str = Field(..., description="Email of the person")
    address: Address = Field(..., description="Address of the person")

pyd_ins = PersonalInfo(
    **{
        "name": "onkar bele",
        "age": 21,
        "email": "onkar.bele@databricks.com",
        "address": {
            "street": "123 Main Street",
            "city": "San Francisco",
            "state": "CA",
            "country": "USA",
            "zip": "94101"
        }
    }
)
print(pyd_ins)
```
