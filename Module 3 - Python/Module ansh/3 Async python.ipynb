# Async Python


```python
import time 
import asyncio
```

```python

async def process():
  print("First step")
  await asyncio.sleep(6)
  print("Second step")

await process()
```

```python

async def process_1():
  print("process_1 - First step")
  await asyncio.sleep(6)
  print("process_1 - Second step")

async def process_2():
  print("process_2 - First step")
  await asyncio.sleep(6)
  print("process_2 - Second step")

await process_1()
await process_2()
```

```python
async def process_1():
  print("process_1 - First step")
  await asyncio.sleep(6)
  result = await process_2()
  print("process_1 - Second step")
  print(f"process_2 result {result}")

async def process_2():
  print("process_2 - First step")
  await asyncio.sleep(6)
  print("process_2 - Second step")
  return "process_2 COMPLETED"

await process_1()
```

Advanced


```python
async def process_1():
  print("process_1 - First step")
  await asyncio.sleep(3)
  print("process_1 - Second step")
  
async def process_2():
  print("process_2 - First step")
  await asyncio.sleep(3)
  print("process_2 - Second step")
 
async def main():

  tasks = await asyncio.gather(process_1(), process_2())
  print("Main Completed")
await main()
```
