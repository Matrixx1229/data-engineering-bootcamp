# Objects and Classes

```python
class employee():

  emp_name = 'Rahul'
  emp_dept = 'IT'
  
  def info(self):
    print(f"Employee {self.emp_name} works for {self.emp_dept}")
    
emp1 = employee()
print(emp1.emp_name)
print(emp1.emp_dept)
print(emp1.info())
```
