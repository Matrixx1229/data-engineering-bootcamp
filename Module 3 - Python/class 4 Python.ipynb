# **Class 8: OOP in Python**

- Data Engineers build reusable tools: readers, transformers, loggers.
- OOP helps you modularize, scale, and reorganize pipeline steps cleanly.
- Instead of repeating code for each file format/API — define a class once, use everywhere.

## Classes & Objects (Core Foundation)

```python
class Customer:
  def __init__(self, name):
    self.name = name
c1 = Customer("Onkar")
print(c1.name)
```

```python
#Build a PipelineStep Class

class PipelineStep:
    def __init__(self, step_name):
        self.step_name = step_name
    def execute(self):
        print(f"Executing {self.step_name}")

ingest =  PipelineStep("Ingestion")
transform = PipelineStep("Transformation")

ingest.execute()
transform.execute()
```

##  Encapsulation – Protect Internal Logic

```python
class DatabaseConnector:
  def __init__(self):
    self._credentials = "user:pass@1234" #protected variable credentials
  
  def connect(self):
    print(f"Connecting to database using {self._credentials}")
    return "DB connection established"
db = DatabaseConnector()
print(db.connect())
```

##  Inheritance – Reuse Code with Variants

```python
class reader():
  def read(self):
    return "Reading data from the base..."
  
class CSVReader(reader):
  def read(self):
    return "Reading data from CSV..."
  
class JSONReader(reader):
  def read(self):
    return "Reading data from JSON..."
  
print(CSVReader().read())
print(JSONReader().read())
```

## Polymorphism – Same Interface, Different Behavior

```python
def run_reader(reader):
  print(reader.read())
run_reader(CSVReader())
run_reader(JSONReader())
```

## Abstraction – Enforce Structure Across Team

```python
from abc import ABC, abstractmethod

class Ingestor(ABC):
  @abstractmethod
  def read(self):
    pass
  
class CSVIngestor(Ingestor):
  def read(self):
    return "Reading data from CSV..."
  
class JSONIngestor(Ingestor):
  def read(self):
    return "Reading data from JSON..."
  
def run_ingestor(ingestor):
  print(ingestor.read())
  
run_ingestor(CSVIngestor())
run_ingestor(JSONIngestor())
```

#  Real Data Pipeline: OOP Version

```python
class FileIngestor(Ingestor):
  def __init__(self, file_path):
    self.file_path = file_path
  
  def read(self):
    print(f"Reading data from {self.file_path}")
    return f"Raw data from {self.file_path}"
  
class cleaner:
  def clean(self, data):
     print("Cleaning data...")
     return f"Cleaned version of: {data}"

class writer:
  def write(self, data):
    print(f"Writing data: {data}")
    return "Write success"
  
#execute data pipeline 
reader = FileIngestor("data.csv")
raw_data = reader.read()

cleaned = cleaner().clean(raw_data)
writer().write(cleaned)
```
